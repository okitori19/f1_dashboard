```python
import fastf1
import pandas as pd

fastf1.Cache.enable_cache('cache')

YEAR = 2025

# COUNTRY = 'Belgium'
COUNTRY = 'Monaco'

# SESSION = 'Practice 1'
# SESSION = 'Practice 2'
# SESSION = 'Practice 3'
# SESSION = 'Qualifying'
SESSION = 'Race'

session = fastf1.get_session(YEAR, COUNTRY, SESSION)
session.load()
```

```python
telemetry_list = []

# Извлекаем ключи из session_info
session_key = session.session_info['Key']
meeting_key = session.session_info['Meeting']['Key']

# Проходим по каждому кругу в сессии
for _, lap in session.laps.iterlaps():
    try:
        # Получаем телеметрию текущего круга
        lap_telemetry = lap.get_telemetry()
        
        # Если телеметрия для круга есть, добавляем информацию о круге и гонщике
        if not lap_telemetry.empty:
            lap_telemetry['MeetingKey'] = meeting_key
            lap_telemetry['SessionKey'] = session_key
            
            lap_telemetry['LapNumber'] = lap['LapNumber']
            lap_telemetry['Driver'] = lap['Driver']
            lap_telemetry['DriverNumber'] = lap['DriverNumber']
            lap_telemetry['IsAccurate'] = lap['IsAccurate']
            lap_telemetry['Compound'] = lap['Compound']  # Можно добавить тип шин
            lap_telemetry['LapTimeSeconds'] = lap['LapTime'].total_seconds()
            lap_telemetry['IsPitOutLap'] = pd.notna(lap['PitOutTime'])
            lap_telemetry['IsPitInLap'] = pd.notna(lap['PitInTime'])
            
            telemetry_list.append(lap_telemetry)
    except Exception as e:
        # Пропускаем круги без данных (например, если нет телеметрии)
        continue

# Объединяем все круги в один общий DataFrame
full_telemetry = pd.concat(telemetry_list, ignore_index=True)

# Сохраняем в CSV
output_filename = f'telemetry_' + str(YEAR) + '_' + COUNTRY + '_' + SESSION + '.csv'
full_telemetry.to_csv(output_filename, index=False)
```

```python
YEAR = 2025

# COUNTRY = 'Belgium'
COUNTRY = 'Monaco'

# SESSION = 'Practice 1'
# SESSION = 'Practice 2'
# SESSION = 'Practice 3'
# SESSION = 'Qualifying'
SESSION = 'Race'

session = fastf1.get_session(YEAR, COUNTRY, SESSION)
session.load()
```

```python
# Получаем данные о поворотах
corners = session.get_circuit_info().corners
# Оставляем только нужные колонки: номер поворота и его дистанцию
corners = corners[['Distance', 'Number']].rename(columns={'Number': 'CornerNumber'})

# ВАЖНО: для merge_asof обе таблицы должны быть отсортированы по ключу (Distance)
full_telemetry = full_telemetry.sort_values('Distance')
corners = corners.sort_values('Distance')


# Объединяем телеметрию с номерами поворотов
# direction='nearest' ищет ближайшую точку. 
# tolerance=30 означает, что номер поворота проставится только для точек 
# в радиусе 30 метров от центра поворота (остальные останутся пустыми / NaN).
telemetry_with_corners = pd.merge_asof(
    full_telemetry, 
    corners, 
    on='Distance', 
    direction='nearest', 
    tolerance=1
)


# Сохраняем в CSV
output_filename = f'telemetry_' + str(YEAR) + '_' + COUNTRY + '_' + SESSION + '.csv'
telemetry_with_corners.to_csv(output_filename, index=False)
```
