```python
import pandas as pd
import numpy as np
import requests


```

```python

# 1. Загрузить данные
df = pd.read_csv('track coordinates.csv')  # <-- укажи свой путь
x = df['x'].values
y = df['y'].values
```

```python

# 2. Считаем приращения и длины отрезков
dx = np.diff(x)
dy = np.diff(y)
ds = np.hypot(dx, dy)


```

```python
# 3. Накопленная длина вдоль трассы
s = np.concatenate(([0], np.cumsum(ds)))

```

```python
# 4. Нормируем в проценты
percent = (s / s[-1]) * 100

```

```python
# 5. Добавляем в DataFrame
df['percent_progress'] = percent
```

```python

# 6. Сохраняем результат
df.to_csv('track_with_percent.csv', index=False)

print(df.head())
```
